```python
# Diseñe una función que calcule la cantidad de carne de aves en kilos si se tienen N gallinas, M gallos y K pollitos cada uno pesando 6 kilos, 7 kilos y 1 kilo respectivamente.
def cantidad_carne(*args):
    N, M, K = args  # desempaquetamos el argumento
    # Pesos de cada tipo de animal en kilos
    peso_gallina = 6
    peso_gallo = 7
    peso_pollito = 1
    total_carne = (N * peso_gallina) + (M * peso_gallo) + (K * peso_pollito) # calculamos la suma de los pesos de cada tipo de animal multiplicando su peso por el numero de animales que hay de cada tipo
    return total_carne # nos retorna el resultado
if __name__ == "__main__":
  n = int(input("Ingrese el número de gallinas: ")) # ingresamos el numero de gallinas
  m = int(input("Ingrese el número de gallos: ")) # ingresamos el numero de gallos
  k = int(input("Ingrese el número de pollitos: ")) # ingresamos el numero de pollitos
  total_carne = cantidad_carne(n, m, k)# Llamamos a la funcion
  print(f"El peso total de los animales es: {total_carne} kilos")   # Imprimimos el resultado de la funcion anterior
```

Escriba una función recursiva para calcular la operación de la potencia.

```python
def pot_recursiva(base, n):
    if n < 0:
        return 1 / pot_recursiva(base, -n) # exponentes negativos
    if n == 0:
        return 1 #exponente igual a 0 el resultado es 1
    return base * pot_recursiva(base, n - 1) # establecemos la condicion recursiva

if __name__ == "__main__":
        base = int(input("Ingrese la base: ")) # se hace el input de la base que sera elevada
        n = int(input("Ingrese el exponente: ")) # la potencia a la cual se eleva la base
        resultado = pot_recursiva(base, n)
        print(str(base) + " elevado a " + str(n) + " es: " + str(resultado)) # se hace la impresion del resultado de manera detallada
```

Realice pruebas para calcular fibonacci con iteración o con recursión. Determine desde que número de la serie la diferencia de tiempo se vuelve significativa.

```python
import time

def fibonacci_recursivo(n):
    if n <= 1:
        return n
    return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2)

def fibonacci_iterativo(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

diferencia_significativa = None

for n in range(30, 41):  # reducimos el rango
    # Medimos el tiempo recursivo
    start_time = time.time()
    fibonacci_recursivo(n)
    end_time = time.time()
    timer_recursivo = end_time - start_time

    # Medimos el tiempo iterativo
    start_time = time.time()
    fibonacci_iterativo(n)
    end_time = time.time()
    timer_iterativo = end_time - start_time

    diferencia = timer_recursivo - timer_iterativo
    print("n = " + str(n) + " Recursivo: " + str(timer_recursivo) + " seg, Iterativo: " + str(timer_iterativo) + " seg, Diferencia: " + str(diferencia) + " seg")

    if not diferencia_significativa and diferencia > 0:
      diferencia_significativa = n

if diferencia_significativa:
    print("La diferencia de tiempo se vuelve significativa a partir de " + str(diferencia_significativa))
else:
    print("No se encontró una diferencia significativa.")

```
